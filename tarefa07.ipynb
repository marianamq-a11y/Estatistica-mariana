## Pergunta 1
Como a acidez das amostras se comporta em relação ao pH? Apresente um
gráfico que permita visualizar essa relação.

Resposta- Os dados mostram alta dispersão e ausência de padrão linear definível. Amostras com valores de pH quase idênticos (ao redor de $2{,}90$) apresentam amplitudes de acidez titulável expressivas, variando de $0{,}45\text{ g/100mL}$ a mais de $0{,}90\text{ g/100mL}$.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Banco de dados
df = pd.read_csv('/content/fisico-quimicas.csv')

# Primeiras linhas e informações das colunas para entender dados
print(df.head())
print(df.info())
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='pH', y='Acidez_g_100mL', data=df)
plt.title('Acidez Total vs. pH')
plt.xlabel('pH')
plt.ylabel('Acidez Total (g/100mL)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## Pergunta 5
O pH pode ser considerado uma variável útil para explicar a acidez das amostras
analisadas? Justifique sua conclusão com base nos resultados.


Resposta- Com base nos resultados da análise, o pH não pode ser considerado uma variável útil para explicar a acidez das amostras analisadas.

O gráfico de dispersão (Acidez_g_100mL vs. pH) mostrou uma alta dispersão dos pontos, sem um padrão linear claro ou forte. Isso indica visualmente que amostras com pH similar podem ter níveis de acidez muito diferentes.

O coeficiente de correlação de Pearson ($r$$r$) foi de 0.1413. Um valor tão próximo de zero indica uma relação linear muito fraca entre o pH e a acidez. Quanto mais próximo de 1 (ou -1), mais forte a correlação linear seria.

O p-valor do teste de correlação foi de 0.2814. Como este p-valor é significativamente maior do que o nível de significância comum ($\alpha = 0.05$$\alpha = 0.05$), não podemos rejeitar a hipótese nula de que não há correlação linear entre pH e acidez. Em outras palavras, a pequena correlação observada pode ser apenas devido ao acaso.

### Questão 1 - Quais valores a variável aleatória $X$ pode assumir?

A variável $X$ representa uma contagem dentro de uma amostra de 20 unidades. Portanto, $X$ pode assumir qualquer valor inteiro de 0 a 20:
   $$X \in \{0, 1, 2, 3, \dots, 20\}$$



### Questão 2 - Qual distribuição de probabilidade pode ser utilizada para representar $X$? Identifique seus parâmetros e justifique sua escolha.

Distribuição: Binomial ($X \sim \text{Binomial}(n = 20, p = 0{,}04)$)   Parâmetros:$n = 20$ (número fixo de ensaios/embalagens)   $p = 0{,}04$ (probabilidade de sucesso/sucesso definido como "estar abaixo do peso")   Justificativa:Existe um número fixo de $n = 20$ ensaios independentes.   Cada embalagem possui apenas dois resultados possíveis: está abaixo do peso ("sucesso") ou não está ("fracasso").   A probabilidade $p = 0{,}04$ permanece constante em cada embalagem.   A função de probabilidade é dada por:$$P(X = k) = \binom{20}{k} (0{,}04)^k (1 - 0{,}04)^{20 - k}$$


### Questão 3 - Qual é a probabilidade de nenhuma das 20 embalagens estar abaixo do peso especificado?

Calculando para $k = 0$:$$P(X = 0) = \binom{20}{0} (0{,}04)^0 (0{,}96)^{20}$$$$P(X = 0) = 1 \cdot 1 \cdot (0{,}96)^{20} \approx 0{,}4420 \quad (44{,}20\%)$$


### Questão 4 - Qual é a probabilidade de exatamente duas embalagens estarem abaixo do peso?

Calculando para $k = 2$:$$P(X = 2) = \binom{20}{2} (0{,}04)^2 (0{,}96)^{18}$$$$\binom{20}{2} = \frac{20 \times 19}{2} = 190$$$$P(X = 2) = 190 \times (0{,}0016) \times (0{,}4796) \approx 0{,}1458 \quad (14{,}58\%)$$


### Questão 5 - Qual é a probabilidade de pelo menos uma embalagem estar abaixo do peso? Existe uma forma mais simples de calcular essa probabilidade utilizando o evento complementar?

Sim, a forma mais simples é utilizando o evento complementar de "pelo menos uma" ($X \ge 1$), que é "nenhuma" ($X = 0$):
   $$P(X \ge 1) = 1 - P(X = 0)$$Aproveitando o resultado da Questão 3 ($P(X = 0) \approx 0{,}4420$):$$P(X \ge 1) = 1 - 0{,}4420 = 0{,}5580 \quad (55{,}80\%)$$


### Questão 6 - Determine $E(X)$. Quantas embalagens abaixo do peso são esperadas, em média, em conjuntos de 20 embalagens avaliadas sob essas condições?

A esperança (ou valor esperado) de uma variável aleatória Binomial é dada por $E(X) = n \cdot p$:$$E(X) = 20 \times 0{,}04 = 0{,}8$$Interpretação: Em média, espera-se encontrar 0,8 embalagem abaixo do peso por lote de 20 embalagens inspecionadas (ou seja, em média menos de 1 embalagem defeituosa por lote analisado).

### Interpretação dos Resultados no Contexto do Controle de Qualidade

A probabilidade de ter ao menos uma embalagem fora da especificação é de 55,80%: isso indica que em mais da metade das inspeções de lotes de 20 unidades haverá ao menos um item inconforme.

O valor esperado de 0,8 indica um processo relativamente sob controle, mas caso o limite tolerado pela regulamentação seja estrito (tolerância zero a itens abaixo do peso líquido), a taxa de defeitos de 4% pode demandar calibração do maquinário de envase para reduzir ainda mais as perdas e penalidades legais.


### Resolução em Python




```python
from scipy.stats import binom

n = 20
p = 0.04

# 3. Probabilidade de NENHUMA estar abaixo do peso P(X = 0)
prob_0 = binom.pmf(0, n, p)

# 4. Probabilidade de EXATAMENTE DUAS P(X = 2)
prob_2 = binom.pmf(2, n, p)

# 5. Probabilidade de PELO MENOS UMA P(X >= 1)
prob_pelo_menos_1 = 1 - binom.pmf(0, n, p)  # ou 1 - prob_0

# 6. Esperança E(X)
media = binom.mean(n, p)

print(f"P(X = 0): {prob_0:.4f}")
print(f"P(X = 2): {prob_2:.4f}")
print(f"P(X >= 1): {prob_pelo_menos_1:.4f}")
print(f"E(X): {media:.1f}")
```

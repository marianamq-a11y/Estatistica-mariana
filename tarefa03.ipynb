```python
import pandas as pd

# Banco de dados para DataFrame
df_alunos = pd.read_excel('/content/Base_Dados_Alimentacao.xlsx')

# Exibir as primeiras 15 linhas do DataFrame
display(df_alunos.head())
```

```python
# 3. Tabelas de frequência

# Variável Nominal: Lanche_Favorito
print("Tabela de Frequência para 'Lanche_Favorito':")
frequencia_lanche = df_alunos['Lanche_Favorito'].value_counts().reset_index()
frequencia_lanche.columns = ['Lanche_Favorito', 'Frequência Absoluta']
frequencia_lanche['Frequência Relativa (%)'] = (frequencia_lanche['Frequência Absoluta'] / len(df_alunos)) * 100
display(frequencia_lanche)

```

```python
# Variável Discreta: Frutas_por_Semana
print("\nTabela de Frequência para 'Frutas_por_Semana':")
frequencia_frutas = df_alunos['Frutas_por_Semana'].value_counts().sort_index().reset_index()
frequencia_frutas.columns = ['Frutas_por_Semana', 'Frequência Absoluta']
frequencia_frutas['Frequência Relativa (%)'] = (frequencia_frutas['Frequência Absoluta'] / len(df_alunos)) * 100
display(frequencia_frutas)

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 4. Gráficos

# Gráfico de Barras para 'Lanche_Favorito'
plt.figure(figsize=(10, 6))
sns.countplot(x='Lanche_Favorito', data=df_alunos, palette='viridis', order=df_alunos['Lanche_Favorito'].value_counts().index)
plt.title('Distribuição dos Lanches Favoritos dos Estudantes')
plt.xlabel('Lanche Favorito')
plt.ylabel('Número de Estudantes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

```

```python
# Histograma para 'Gasto_FastFood_RS'
plt.figure(figsize=(10, 6))
sns.histplot(df_alunos['Gasto_FastFood_RS'], bins=5, kde=False, color='skyblue')
plt.title('Distribuição do Gasto Mensal com Fast Food (R$)')
plt.xlabel('Gasto Mensal com Fast Food (R$)')
plt.ylabel('Frequência')
plt.show()

```

```python
# Gráfico de Dispersão para 'Frutas_por_Semana' vs. 'Gasto_FastFood_RS'
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Frutas_por_Semana', y='Gasto_FastFood_RS', data=df_alunos, hue='Lanche_Favorito', size='Gasto_FastFood_RS', sizes=(50, 500), alpha=0.7)
plt.title('Relação entre Consumo de Frutas e Gasto com Fast Food')
plt.xlabel('Frutas por Semana')
plt.ylabel('Gasto Mensal com Fast Food (R$)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

```

**Interpretação da Tabela de Frequência (Lanche Favorito)**
Ao analisar a tabela, podemos observar a contagem absoluta e a proporção das preferências alimentares da turma. Fica evidente que a Pizza e o Hambúrguer são as opções mais populares, concentrando a maior parte das escolhas dos estudantes avaliados.


**Interpretação da Tabela de Frequência (Frutas por Semana)**
Ao analisar esta tabela, que detalha o consumo semanal de frutas (nossa variável discreta), identificamos uma forte concentração de dados na faixa de baixo consumo. Observando as frequências absoluta e relativa, notamos que a maior parte da turma apresenta um baixo consumo de frutas: 20% dos estudantes (3 alunos) não consomem nenhuma porção semanal, e outros 20% consomem apenas duas.

Se somarmos os alunos que comem de zero a duas frutas por semana, temos mais da metade da nossa amostra. Em contrapartida, os estudantes com maior ingestão de frutas (aquelas que consomem 10, 12 ou 14 porções na semana) são casos bem isolados no grupo, representando apenas cerca de 6,6% (1 aluno) para cada uma dessas quantidades.

**Interpretação do Gráfico de Barras**
Este gráfico facilita a visualização dos dados da tabela de frequência. As barras mais altas confirmam rapidamente a predominância da Pizza e do Hambúrguer como os favoritos do grupo, permitindo uma comparação visual direta com as opções menos frequentes.

**Interpretação do Histograma (Gastos com Fast-food)**

Além disso, é possível identificar alguns comportamentos fora desse padrão central, com uma minoria gastando menos de 50 reais

**Interpretação do Gráfico de Dispersão**
Utilizamos este gráfico para investigar a relação entre o consumo semanal de frutas (eixo X) e o valor gasto com fast-food (eixo Y). Observamos uma tendência de correlação negativa: de modo geral, os estudantes que consomem menos frutas tendem a apresentar maiores gastos com fast-food, e vice-versa. Embora a dispersão mostre que a relação não é perfeitamente linear para todos os indivíduos, o padrão de comportamento da turma é perceptível.

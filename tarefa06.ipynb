# Regressão Linear Simples e Correlação Linear de Pearson

## Relação entre massa e volume de leite

## 1. Apresentação do conjunto de dados

Neste trabalho, será analisada a relação entre a massa de leite, expressa em gramas, e o volume de leite, expresso em mililitros.

Os dados utilizados foram criados para fins didáticos, simulando diferentes porções de leite. Espera-se observar uma relação aproximadamente linear entre as duas variáveis.

A massa de leite será considerada a variável independente (X), enquanto o volume de leite será considerado a variável dependente (Y).

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
import pandas as pd

# Link direto para o seu arquivo do Excel no GitHub
url = "https://raw.githubusercontent.com/marianamq-a11y/Estatistica-mariana/main/dados_semana%2006.xlsx"

# Leitura direta no Pandas
dados = pd.read_excel(
    url,
    header=6,
    usecols="G:H",
    names=["Massa", "Volume"]
)

dados
```

```python
import matplotlib.pyplot as plt

plt.scatter(dados["Massa"], dados["Volume"])
plt.xlabel("Massa de leite (g)")
plt.ylabel("Volume de leite (mL)")
plt.title("Relação entre massa e volume do leite")
plt.show()
```

### Regressão linear simples

A regressão linear simples é um método estatístico utilizado para analisar a relação entre duas variáveis numéricas. Ela permite estimar uma variável em função de outra por meio de uma reta.

Neste estudo, a massa (g) é a variável independente (x) e o volume (mL) é a variável dependente (y).

A equação da reta de regressão linear simples é:

y = a + bx

Em que:

* y = valor estimado da variável dependente;
* x = valor da variável independente;
* a = intercepto da reta;
* b = coeficiente angular da reta.


```python
import numpy as np

x = dados["Massa"]
y = dados["Volume"]

b, a = np.polyfit(x, y, 1)

print("Coeficiente angular (b) =", b)
print("Intercepto (a) =", a)
```

```python
y_pred = a + b * x

plt.scatter(x, y, label="Dados observados")
plt.plot(x, y_pred, color="red", label="Reta estimada")
plt.xlabel("Massa de leite (g)")
plt.ylabel("Volume de leite (mL)")
plt.title("Relação entre massa e volume do leite com Reta de Regressão")
plt.legend()
plt.show()
```

### Interpretação dos coeficientes

Para os dados analisados, foram obtidos os seguintes coeficientes:

* Coeficiente angular (b) = 0,9715
* Intercepto (a) = 0,0663

O coeficiente angular indica que, para cada aumento de 1 g na massa, espera-se um aumento de aproximadamente 0,9715 mL no volume.

O intercepto indica o valor estimado do volume quando a massa é igual a zero. Neste caso, o valor obtido foi de aproximadamente 0,0663 mL.


### Correlação linear de Pearson

A correlação linear de Pearson é uma medida estatística que indica a intensidade e a direção da relação linear entre duas variáveis numéricas. Seu valor varia de -1 a 1.

A fórmula da correlação de Pearson é:

r = Σ[(xᵢ − x̄)(yᵢ − ȳ)] / √[Σ(xᵢ − x̄)² · Σ(yᵢ − ȳ)²]

Em que:

* r = coeficiente de correlação de Pearson;
* xᵢ e yᵢ = valores observados das duas variáveis;
* x̄ e ȳ = médias das variáveis.

Valores de "r" próximos de 1 indicam uma correlação linear positiva forte, enquanto valores próximos de -1 indicam uma correlação linear negativa forte. Valores próximos de zero indicam ausência ou pouca relação linear.


```python
r = x.corr(y)

print("Correlação de Pearson (r) =", r)
```

### Interpretação da correlação

A correlação de Pearson obtida foi:

r = 0,99999916

Esse resultado indica uma correlação linear positiva extremamente forte, praticamente perfeita, entre massa e volume. Isso significa que, à medida que a massa aumenta, o volume também aumenta de maneira aproximadamente linear.

É importante destacar que uma correlação igual ou próxima de zero indica ausência de relação linear, mas não necessariamente ausência de qualquer relação entre as variáveis.


### Conclusão

A análise dos dados mostrou uma relação linear positiva extremamente forte entre massa e volume. A regressão linear simples permitiu estimar essa relação por meio de uma reta, enquanto o coeficiente de correlação de Pearson apresentou valor de 0,99999916, indicando uma correlação positiva praticamente perfeita.

Os resultados mostram que, nos dados analisados, o aumento da massa está associado ao aumento do volume de forma aproximadamente linear. Dessa forma, a regressão linear e a correlação de Pearson são ferramentas úteis para analisar e interpretar relações entre variáveis numéricas na área de Ciências dos Alimentos.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the frequency of each fruit preference
fruit_counts = df['Qual dessas frutas você prefere para suco natural?'].value_counts().reset_index()
fruit_counts.columns = ['Fruta', 'Contagem']

# Create the bar plot
plt.figure(figsize=(10,6))
sns.barplot(x='Fruta', y='Contagem', data=fruit_counts, palette='viridis')
plt.title('Preferência de Frutas para Suco Natural')
plt.xlabel('Fruta')
plt.ylabel('Número de Respostas')
plt.xticks(rotation=90, ha='right') # Rotate labels for better readability
plt.tight_layout()
plt.show()
```

# Tipos de variáveis

##Qual. nominal
1- Estado civil (solteiro, casado, divorciado...)

2- País onde mora (Brasil, Italia, Chile...)

3- Tipo de iogurte (Morango, natural, coco...)

##Qual. ordinal
1- Qualidade do produto (ruim, bom ,ótimo...)

2- Intensidade do café (forte, fraco, médio...)

3- Grau de satisfação (baixo, médio, alto...)

##Quant. Discreta
1- Numero de plantas por vaso (0, 1, 2...)

2- Numero de alunos por disciplina (0, 1, 2...)

3- Numero de placas de petri conservadas (0, 1, 2...)

##Quant. Continua
1- Teor de sólidos soluveis (10%, 20%, 30%...)

2- Peso por amostra (2g, 10g, 15g...)

3- Volume de litros de leite produzido por vaca (20L, 30L, 40L)

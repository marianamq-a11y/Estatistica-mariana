### Questão 1 - Escreva a distribuição de probabilidade de 𝑋 , identificando seus parâmetros.

A variável aleatória $X$ segue uma distribuição Normal com média $\mu = 12{,}5$ e variância $\sigma^2 = 0{,}64$:
   $$X \sim N(\mu = 12{,}5;\, \sigma^2 = 0{,}64) \quad \text{ou} \quad X \sim N(12{,}5;\, 0{,}8^2)$$
   
   Parâmetros:Média ($\mu$): $12{,}5$   
   
   Desvio-padrão ($\sigma$): $0{,}8$


### Questão 2 - Qual é a probabilidade de uma unidade apresentar teor de proteína entre 11,5 e 13,5g/100 g? Escreva a probabilidade na forma 𝑃(11,5 ≤ 𝑋 ≤ 13,5).



$$Z_1 = \frac{11{,}5 - 12{,}5}{0{,}8} = -1{,}25 \quad \text{e} \quad Z_2 = \frac{13{,}5 - 12{,}5}{0{,}8} = 1{,}25$$Cálculo da probabilidade:$$P(11{,}5 \le X \le 13{,}5) = P(-1{,}25 \le Z \le 1{,}25) = \Phi(1{,}25) - \Phi(-1{,}25)$$

$$P(11{,}5 \le X \le 13{,}5) = 0{,}8944 - 0{,}1056 = \mathbf{0{,}7888} \quad (\text{ou } 78{,}88\%)$$


### Questão 3 - Represente graficamente a distribuição de 𝑋 e destaque a região correspondente ao intervalo considerado.



```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Parâmetros da distribuição
mu = 12.5
sigma = 0.8

# Valores para o eixo X
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = norm.pdf(x, mu, sigma)

# Criando o gráfico
plt.figure(figsize=(8, 5))
plt.plot(x, y, "b-", linewidth=2, label="Distribuição Normal")

# Sombreando a região entre 11.5 e 13.5
x_fill = np.linspace(11.5, 13.5, 500)
y_fill = norm.pdf(x_fill, mu, sigma)
plt.fill_between(
    x_fill,
    y_fill,
    color="skyblue",
    alpha=0.5,
    label="P(11.5 ≤ X ≤ 13.5) ≈ 78.88%",
)

# Marcações e destaques
plt.axvline(12.5, color="gray", linestyle="--", label="Média (μ = 12.5)")
plt.axvline(11.5, color="red", linestyle=":")
plt.axvline(13.5, color="red", linestyle=":")

# Títulos e rótulos
plt.title("Distribuição do Teor de Proteína - Questão 3", fontsize=14)
plt.xlabel("Teor de Proteína (g/100 g)", fontsize=12)
plt.ylabel("Densidade de Probabilidade", fontsize=12)
plt.legend(loc="upper right")
plt.grid(True, linestyle=":", alpha=0.6)

plt.show()
```

### Questão 4 - Qual é a probabilidade de uma unidade apresentar teor de proteína superior a 14 g/100 g?


$$Z = \frac{14 - 12{,}5}{0{,}8} = 1{,}875 \approx 1{,}88$$Cálculo da probabilidade:$$P(X > 14) = P(Z > 1{,}88) = 1 - \Phi(1{,}88) = 1 - 0{,}9699 = \mathbf{0{,}0301} \quad (\text{ou } 3{,}01\%)$$


### Questão 5 - Represente graficamente a probabilidade calculada no item anterior.


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Parâmetros da distribuição
mu = 12.5
sigma = 0.8

# Valores para o eixo X
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = norm.pdf(x, mu, sigma)

# Criando o gráfico
plt.figure(figsize=(8, 5))
plt.plot(x, y, "b-", linewidth=2, label="Distribuição Normal")

# Sombreando a região superior a 14 (cauda direita)
x_fill = np.linspace(14, mu + 4 * sigma, 500)
y_fill = norm.pdf(x_fill, mu, sigma)
plt.fill_between(
    x_fill, y_fill, color="orange", alpha=0.5, label="P(X > 14) ≈ 3.01%"
)

# Marcações e destaques
plt.axvline(12.5, color="gray", linestyle="--", label="Média (μ = 12.5)")
plt.axvline(14.0, color="red", linestyle=":", label="X = 14.0")

# Títulos e rótulos
plt.title("Distribuição do Teor de Proteína - Questão 5", fontsize=14)
plt.xlabel("Teor de Proteína (g/100 g)", fontsize=12)
plt.ylabel("Densidade de Probabilidade", fontsize=12)
plt.legend(loc="upper right")
plt.grid(True, linestyle=":", alpha=0.6)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregando o arquivo Excel
file_path = '/content/dados_semana02.xlsx'
df = pd.read_excel(file_path)

# Exibindo as primeiras linhas e informações das colunas
print("Colunas disponíveis:", df.columns.tolist())
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurações de estilo
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Gráfico de Pizza (Variável: Sexo)
sexo_counts = df['Sexo'].value_counts()
axes[0, 0].pie(sexo_counts, labels=sexo_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
axes[0, 0].set_title('Distribuição por Sexo')

# 2. Gráfico de Barras (Variável: Classificação IMC)
sns.countplot(data=df, x='Classificação IMC', ax=axes[0, 1], palette='viridis', hue='Classificação IMC', legend=False)
axes[0, 1].set_title('Frequência por Classificação de IMC')
axes[0, 1].set_xlabel('Classificação')
axes[0, 1].set_ylabel('Quantidade')
axes[0, 1].tick_params(axis='x', rotation=45)

# 3. Boxplot (Variáveis: Nível de Atividade e Idade)
sns.boxplot(data=df, x='Nível de Atividade', y='Idade (anos)', ax=axes[1, 0], palette='Set2', hue='Nível de Atividade', legend=False)
axes[1, 0].set_title('Boxplot de Idade por Nível de Atividade')
axes[1, 0].set_xlabel('Nível de Atividade')
axes[1, 0].set_ylabel('Idade (anos)')

# 4. Histograma (Variável: IMC)
sns.histplot(df['IMC'], kde=True, ax=axes[1, 1], color='skyblue')
axes[1, 1].set_title('Distribuição de IMC (Histograma)')
axes[1, 1].set_xlabel('IMC')
axes[1, 1].set_ylabel('Frequência')

plt.tight_layout()
plt.show()
```

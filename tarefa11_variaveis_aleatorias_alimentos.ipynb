# 1. Variável Aleatória

Uma variável aleatória é uma função que associa um valor numérico a cada possível resultado de um experimento aleatório. Ela permite representar matematicamente resultados que não podem ser determinados antecipadamente.

Nas Ciências dos Alimentos, uma variável aleatória pode ser utilizada para representar características ou ocorrências observadas durante a produção e o controle de qualidade de alimentos.

Por exemplo, em um lote de embalagens de um produto alimentício, pode-se realizar uma inspeção para identificar quantas embalagens apresentam defeitos. Nesse caso, podemos definir a variável aleatória **X** como o número de embalagens defeituosas encontradas no lote.

Assim, os valores possíveis de X dependem da quantidade de embalagens avaliadas. Por exemplo, se forem avaliadas 20 embalagens, X poderá assumir valores de 0 a 20.


```python
# Exemplo de variável aleatória:
# número de embalagens defeituosas encontradas em um lote

numero_embalagens = 20

print("Quantidade de embalagens avaliadas:", numero_embalagens)
print("A variável aleatória X representa o número de embalagens defeituosas.")
print("Valores possíveis de X: de 0 a", numero_embalagens)
```

# 2. Variáveis Aleatórias Discretas e Contínuas

As variáveis aleatórias podem ser classificadas em discretas ou contínuas.

As **variáveis aleatórias discretas** são aquelas que assumem valores que podem ser contados, geralmente números inteiros. Na área de Ciências dos Alimentos, alguns exemplos são:

* **Número de embalagens reprovadas em um lote:** é uma variável discreta porque representa uma contagem e pode assumir valores inteiros.
* **Número de reclamações de consumidores em um dia:** é discreta porque as reclamações são contadas individualmente.
* **Número de frutas danificadas em uma caixa:** é discreta porque representa a quantidade de frutas que apresentam danos.

As **variáveis aleatórias contínuas** podem assumir qualquer valor dentro de determinado intervalo, incluindo valores decimais. Exemplos relacionados às Ciências dos Alimentos são:

* **Massa de um produto alimentício:** é contínua porque pode assumir valores como 499,8 g, 500,0 g ou 500,3 g.
* **Temperatura de armazenamento:** é contínua porque pode assumir valores decimais, como 4,2 °C ou 4,7 °C.
* **Teor de umidade de um alimento:** é contínua porque pode apresentar diferentes valores dentro de um intervalo, incluindo valores decimais.

Portanto, a principal diferença é que as variáveis discretas estão associadas principalmente à contagem de ocorrências, enquanto as variáveis contínuas estão relacionadas a medidas que podem assumir infinitos valores dentro de um intervalo.


# 3. Modelo de Bernoulli

O modelo de Bernoulli é utilizado em experimentos que apresentam apenas dois resultados possíveis. Esses resultados podem ser representados como sucesso e fracasso.

Como exemplo na área de Ciências dos Alimentos, pode-se considerar a inspeção da selagem de uma embalagem.

Neste exemplo:

* **Sucesso:** embalagem apresenta selagem adequada;
* **Fracasso:** embalagem apresenta selagem inadequada.

A variável aleatória pode ser representada por **X**, assumindo:

* X = 1, quando a embalagem apresenta selagem adequada;
* X = 0, quando a embalagem apresenta selagem inadequada.

Esse modelo representa apenas **um ensaio**, ou seja, a avaliação de uma única embalagem.


```python
from scipy.stats import bernoulli

# Parâmetro do modelo: 90% de probabilidade de selagem adequada (p = 0.90)
p_sucesso = 0.90
dist_bernoulli = bernoulli(p=p_sucesso)

# Probabilidades teóricas do ensaio de Bernoulli
prob_sucesso = dist_bernoulli.pmf(1)  # P(X = 1) -> Selagem adequada
prob_fracasso = dist_bernoulli.pmf(0)  # P(X = 0) -> Selagem inadequada

print("--- MODELO DE BERNOULLI ---")
print(f"Probabilidade de selagem adequada P(X=1): {prob_sucesso*100:.1f}%")
print(f"Probabilidade de selagem inadequada P(X=0): {prob_fracasso*100:.1f}%")

# Simulação da inspeção em 10 embalagens
simulacao = dist_bernoulli.rvs(size=10, random_state=42)

print("\n--- SIMULAÇÃO DE INSPEÇÃO (10 EMBALAGENS) ---")
print("Resultados obtidos (1 = adequada, 0 = inadequada):")
print(simulacao)
```

# 4. Modelo Binomial

O modelo Binomial é utilizado para representar o número de sucessos obtidos em um número fixo de ensaios independentes. Para sua aplicação, cada ensaio deve apresentar apenas dois resultados possíveis, sucesso ou fracasso, e a probabilidade de sucesso deve permanecer constante.

Na área de Ciências dos Alimentos, o modelo pode ser aplicado à inspeção da selagem de embalagens de um produto alimentício. Nesse caso, cada embalagem pode apresentar selagem adequada ou inadequada.

Considerando uma amostra de 20 embalagens, podemos utilizar o modelo Binomial para determinar a probabilidade de obter determinada quantidade de embalagens com selagem adequada.

Neste exemplo:

* **Número de ensaios (n):** 20 embalagens;
* **Sucesso:** embalagem com selagem adequada;
* **Fracasso:** embalagem com selagem inadequada;
* **Probabilidade de sucesso (p):** 0,90;
* **Variável aleatória (X):** número de embalagens com selagem adequada entre as 20 avaliadas.

A probabilidade de sucesso de 90% é um valor **simulado**, utilizado para demonstrar a aplicação do modelo em uma situação hipotética de controle de qualidade.


```python
from scipy.stats import binom

# Parâmetros da distribuição Binomial
n = 20       # número de embalagens avaliadas
p = 0.90     # probabilidade de selagem adequada

# Número de sucessos desejado
x = 18

# Cálculo da probabilidade
probabilidade = binom.pmf(x, n, p)

print(f"Probabilidade de exatamente {x} embalagens apresentarem selagem adequada:")
print(f"{probabilidade:.4f}")
print(f"Ou aproximadamente {probabilidade * 100:.2f}%")
```

```python
# Probabilidade de pelo menos 18 embalagens apresentarem
# selagem adequada

probabilidade_18_ou_mais = binom.sf(17, n, p)

print("Probabilidade de pelo menos 18 embalagens apresentarem selagem adequada:")
print(f"{probabilidade_18_ou_mais:.4f}")
print(f"Ou aproximadamente {probabilidade_18_ou_mais * 100:.2f}%")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Valores possíveis para X
x = np.arange(0, n + 1)

# Probabilidade para cada valor de X
probabilidades = binom.pmf(x, n, p)

# Gráfico
plt.figure(figsize=(10, 5))
plt.bar(x, probabilidades)

plt.xlabel("Número de embalagens com selagem adequada")
plt.ylabel("Probabilidade")
plt.title("Distribuição Binomial - Inspeção de 20 embalagens")

plt.show()
```

### Interpretação

No exemplo apresentado, foram consideradas 20 embalagens, sendo que cada uma possui probabilidade de 90% de apresentar selagem adequada. O modelo Binomial permite calcular a probabilidade de diferentes quantidades de embalagens apresentarem o resultado considerado como sucesso.

A probabilidade de exatamente 18 embalagens apresentarem selagem adequada foi de aproximadamente 28,52%. Além disso, a probabilidade de pelo menos 18 das 20 embalagens apresentarem selagem adequada foi de aproximadamente 67,69%.

O modelo Binomial é adequado para esse exemplo porque existe um número fixo de ensaios, cada ensaio possui dois resultados possíveis e foi considerada uma probabilidade constante de sucesso.


# 5. Modelo de Poisson

   5.1 Conceito e Características
   
   O Modelo de Poisson é uma distribuição de probabilidade discreta empregada para modelar o número de ocorrências de um evento de interesse em um intervalo contínuo fixo (tempo, área, volume, massa ou unidade de produção).
   
   Diferente do modelo Binomial (onde há um número fixo de ensaios $n$), no modelo de Poisson o número máximo de ocorrências $x$ não possui um limite teórico superior fixo ($x \in \{0, 1, 2, 3, \dots\}$).
   
   Condições de Aplicação (Processo de Poisson):
   
   Taxa Média Constante ($\lambda$): O evento ocorre a uma taxa média conhecida ($\lambda > 0$) no intervalo considerado.
   
   Independência: A ocorrência de um evento em um subintervalo é independente da ocorrência em outro subintervalo disjunto.
   
   Simultaneidade Negligenciável: A probabilidade de duas ou mais ocorrências simultâneas em um intervalo infinitesimalmente pequeno é praticamente zero.
   
   Função de Massa de Probabilidade (FMP):$$P(X = x) = \frac{e^{-\lambda} \cdot \lambda^x}{x!}$$Onde:$X$: Variável aleatória discreta (contagem de eventos).
   
   $x$: Número de ocorrências desejado ($x = 0, 1, 2, \dots$).
   
   $\lambda$ (lambda): Taxa média de eventos no intervalo de interesse.
   
   $e$: Base do logaritmo natural ($\approx 2,71828$).

```python
# Importação de bibliotecas para o Modelo de Poisson e Comparação
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Configuração de estilo visual dos gráficos
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# ==============================================================================
# EXEMPLO CONTEXTUALIZADO: Contagem de Colônias de Bolores e Leveduras
# Contexto: Inspeção de Qualidade Microbiológica no Armazenamento de Queijos
# ==============================================================================
# Em uma queijaria, o controle de qualidade realiza contagens de Unidades Formadoras
# de Colônias (UFC) de bolores e leveduras em placas de amostragem por superfície (100 cm²).
# Com base no histórico de monitoramento ambiental do armazém de maturação,
# sabe-se que a taxa média é de lambda = 3 UFC / 100 cm².

lambda_ufc = 3.0

print("--- MODELO DE POISSON: ANÁLISE MICROBIOLÓGICA DE SUPERFÍCIE ---")
print(f"Taxa média estimada (lambda): {lambda_ufc} UFC por 100 cm²\n")

# A) Probabilidade de encontrar exatamente 0 UFC (placa completamente limpa)
p_x0 = poisson.pmf(k=0, mu=lambda_ufc)

# B) Probabilidade de encontrar exatamente 3 UFC (igual à média)
p_x3 = poisson.pmf(k=3, mu=lambda_ufc)

# C) Probabilidade de encontrar mais de 5 UFC (limite crítico de contaminação)
# P(X > 5) = 1 - P(X <= 5)
p_x_gt5 = 1 - poisson.cdf(k=5, mu=lambda_ufc)

print(f"a) P(X = 0 UFC):  {p_x0:.4f} ({p_x0*100:.2f}%)")
print(f"b) P(X = 3 UFC):  {p_x3:.4f} ({p_x3*100:.2f}%)")
print(f"c) P(X > 5 UFC) [Alerta do limite crítico]: {p_x_gt5:.4f} ({p_x_gt5*100:.2f}%)\n")

# Visualização da Distribuição de Poisson
k_valores = np.arange(0, 12)
pmf_poisson = poisson.pmf(k_valores, mu=lambda_ufc)

plt.figure(figsize=(9, 5))
bars = plt.bar(k_valores, pmf_poisson, color='#2c3e50', alpha=0.85, edgecolor='black', linewidth=1.2)

# Destacar barra do limite crítico (> 5)
for i in range(6, len(k_valores)):
    bars[i].set_color('#e74c3c')

plt.title(f'Distribuição de Poisson - Contagem de UFC ($\lambda = {lambda_ufc}$ UFC/100cm²)', fontsize=13, fontweight='bold')
plt.xlabel('Número de Colônias Observadas (UFC)', fontsize=11)
plt.ylabel('Probabilidade P(X = x)', fontsize=11)
plt.xticks(k_valores)
plt.axvline(x=5.5, color='red', linestyle='--', label='Limite Crítico de Alerta (> 5 UFC)')
plt.legend(loc='upper right')
plt.grid(axis='y', alpha=0.3)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.005, f'{yval:.3f}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

5.2 Interpretação dos Resultados Microbiológicos

A probabilidade de obter uma amostra inteiramente isenta de contaminação visível ($X = 0$) é de 4,98%.

A maior probabilidade individual ocorre exatamente nos valores 2 e 3 UFC (22,40% cada), alinhando-se com a taxa média $\lambda = 3$.

A probabilidade de superação do limite crítico de segurança do armazém ($X > 5\text{ UFC}$) é de 8,39%. Esse resultado orienta a equipe de garantia da qualidade sobre a frequência ideal de sanificação das prateleiras de maturação.
